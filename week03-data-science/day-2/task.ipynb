| **Engineered Feature**                               | **Hypothesis**                                                                                      | **Leakage Risk**                                                     |
| ---------------------------------------------------- | --------------------------------------------------------------------------------------------------- | -------------------------------------------------------------------- |
| **AgeGroup** (created using `pd.cut(Age)`)           | Customers in different age groups have different purchasing preferences and spending habits.        | **No** – Age is known before prediction.                             |
| **PriceCategory** (Low, Medium, High, Premium)       | Products in higher price categories may generate higher revenue but lower purchase frequency.       | **No** – Product price is available before purchase.                 |
| **Year**                                             | Customer behavior may change across different years due to market trends or economic conditions.    | **No** – Order year is available when the order is placed.           |
| **Month**                                            | Sales may increase during festive or holiday months, making month a useful predictor of demand.     | **No** – Month is known from the order date.                         |
| **Quarter**                                          | Different quarters may show seasonal buying patterns (e.g., Q4 holiday shopping).                   | **No** – Derived from the order date.                                |
| **Weekday**                                          | Customers may purchase different products depending on the day of the week.                         | **No** – Known at prediction time.                                   |
| **Weekend**                                          | Weekend customers may spend more or buy different product categories than weekday customers.        | **No** – Derived from the order date.                                |
| **TotalAmount** (`Price × Quantity`)                 | Customers with higher order values may be more valuable and less likely to cancel or return orders. | **No** – Safe if quantity is known at the time of prediction.        |
| **DiscountRatio** (`Discount / 100`)                 | Higher discounts may increase purchase probability but could reduce profit margins.                 | **No** – Safe if the discount is determined before the purchase.     |
| **FinalPrice** (`TotalAmount × (1 - DiscountRatio)`) | The final amount paid may better represent customer spending than the original price.               | **Low Risk** – Safe only if the discount is known before prediction. |
| **AgePriceInteraction** (`Age × Price`)              | Customers of different age groups may spend differently on expensive products.                      | **No** – Uses only available features.                               |
| **DiscountImpact** (`Discount × Quantity`)           | Large discounts on bulk purchases may encourage higher sales volumes.                               | **No** – Safe if both values are available before prediction.        |


| **Feature**                                     | **Leakage Risk** | **Reason**                                                                                                                             |
| ----------------------------------------------- | ---------------- | -------------------------------------------------------------------------------------------------------------------------------------- |
| **Returned** (Target Variable)                  | ❌ High           | This is usually the value you are trying to predict. Using it as an input gives the model the answer.                                  |
| **ReturnReason**                                | ❌ High           | Available only after the product has been returned.                                                                                    |
| **CustomerReview**                              | ❌ High           | Reviews are written after delivery, so they reveal future information.                                                                 |
| **FuturePurchaseCount**                         | ❌ High           | Uses customer actions that happen after the prediction date.                                                                           |
| **ShipmentStatus**                              | ❌ High           | Shipping information is not available when predicting at order placement.                                                              |
| **DeliveryDays**                                | ⚠️ Depends       | Safe only if you're predicting after shipment; otherwise it leaks future information because the actual delivery time isn't known yet. |
| **CustomerLifetimeValue (using future orders)** | ❌ High           | Includes purchases that haven't happened at prediction time.                                                                           |
