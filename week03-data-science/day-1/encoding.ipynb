Coding


```python
import pandas as pd

df = pd.DataFrame({
    "CustomerID":[101,102,103,104,105],
    "Membership":["Silver","Gold","Bronze","Gold","Silver"]
})

print(df)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoder = LabelEncoder()
```

```python
df["Membership_Encoded"] = encoder.fit_transform(df["Membership"])
```

```python
print(df)
```

```python
mapping = dict(zip(encoder.classes_,
                   encoder.transform(encoder.classes_)))

print(mapping)
```

One hot encoding 

```python
import pandas as pd

df = pd.DataFrame({
    "CustomerID":[101,102,103,104,105],
    "PaymentMethod":[
        "Cash",
        "UPI",
        "Card",
        "Cash",
        "UPI"
    ]
})

print(df)
```

```python
df_encoded = pd.get_dummies(
    df,
    columns=["PaymentMethod"],
    dtype=int
)
```

```python
print(df_encoded)
```
