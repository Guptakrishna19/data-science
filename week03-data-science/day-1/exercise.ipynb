Handling the missing values 

```python
df.isnull().sum()
```

```python
df["Category"] = df["Category"].fillna(df["Category"].mode())
```

```python
df["PaymentMethod"] = df["PaymentMethod"].fillna("cash")
```

```python
df["Price"] = df["Price"].astype(str).str.replace(r'[^0-9.]', '', regex=True)
df["Price"] = pd.to_numeric(df["Price"], errors='coerce')
```

```python
df["Price"] = df["Price"].fillna(df["Price"].mean())
```

```python
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].mean())
```

```python
df["Rating"] = df["Rating"].fillna(df["Rating"].mean())
```

```python
df["OrderDate"] = pd.to_datetime(df["OrderDate"], format="mixed", errors="coerce")
```

```python
df = df.dropna(subset=["CustomerName", "OrderDate"])
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df = df.drop_duplicates()
```

```python
df.dtypes
```

```python
df["CustomerName"] = df["CustomerName"].str.lower()
```

```python
df["CustomerName"] = df["CustomerName"].str.strip()
```

```python
df["CustomerName"] = df["CustomerName"].str.title()
```

```python
df["Category"] = df["Category"].str.lower()
df["Category"] = df["Category"].str.strip()
df["Category"] = df["Category"].str.title()
```

```python
df["PaymentMethod"] = df["PaymentMethod"].str.lower()
df["PaymentMethod"] = df["PaymentMethod"].str.strip()
df["PaymentMethod"] = df["PaymentMethod"].str.title()
```

```python
df["OrderStatus"] = df["OrderStatus"].str.lower()
df["OrderStatus"] = df["OrderStatus"].str.strip()
df["OrderStatus"] = df["OrderStatus"].str.title()
```

Outliers

```python
Q1 = df["Price"].quantile(0.25)
Q3 = df["Price"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR
outliers = df[
    (df["Price"] < lower) |
    (df["Price"] > upper)
]
df = df[
    (df["Price"] >= lower) &
    (df["Price"] <= upper)]
```

```python
Q1 = df["Quantity"].quantile(0.25)
Q3 = df["Quantity"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR
outliers = df[
    (df["Quantity"] < lower) |
    (df["Quantity"] > upper)
]
df = df[
    (df["Quantity"] >= lower) &
    (df["Quantity"] <= upper)]
```

```python
Q1 = df["Rating"].quantile(0.25)
Q3 = df["Rating"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR
outliers = df[
    (df["Rating"] < lower) |
    (df["Rating"] > upper)
]
df = df[
    (df["Rating"] >= lower) &
    (df["Rating"] <= upper)]
```

```python
from scipy.stats import zscore

z = zscore(df["Price"])

df = df[abs(z) < 3]
```

```python
from scipy.stats import zscore

z = zscore(df["Quantity"])

df = df[abs(z) < 3]
```

```python
from scipy.stats import zscore

z = zscore(df["Rating"])

df = df[abs(z) < 3]
```

```python
df["Price"] = pd.to_numeric(
    df["Price"],
    errors="coerce"
)
```

Label encoding 

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df["Category"] = encoder.fit_transform(df["Category"])
```

one hot encoding 

```python
df = pd.get_dummies(
    df,
    columns=["PaymentMethod", "OrderStatus"],
    drop_first=True
)
```

```python
df.to_csv(
    "ecommerce_cleaned.csv",
    index=False
)
```
