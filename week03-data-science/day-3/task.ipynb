| No. | Chart Type       | Use When...                                                                         | Avoid When...                                      | Example                                       |
| --- | ---------------- | ----------------------------------------------------------------------------------- | -------------------------------------------------- | --------------------------------------------- |
| 1   | **Line Chart**   | Showing trends over time (daily, monthly, yearly sales, stock prices, temperature). | Comparing unrelated categories.                    | Monthly Sales Trend                           |
| 2   | **Bar Chart**    | Comparing values across categories (products, cities, departments).                 | Displaying continuous data distributions.          | Sales by Product Category                     |
| 3   | **Histogram**    | Understanding the distribution of a numerical variable (age, salary, marks).        | Comparing categorical values.                      | Distribution of Customer Bills                |
| 4   | **Box Plot**     | Detecting outliers, spread, and comparing distributions between groups.             | Showing exact values for every observation.        | Tips by Day                                   |
| 5   | **Scatter Plot** | Exploring relationships or correlations between two numerical variables.            | Showing time-series trends.                        | Total Bill vs Tip                             |
| 6   | **Heatmap**      | Visualizing correlations, intensity, or matrix values.                              | Showing precise individual values.                 | Correlation Matrix                            |
| 7   | **Pairplot**     | Exploring relationships among multiple numerical variables simultaneously.          | Very large datasets (slow and cluttered).          | Relationships among Total Bill, Tip, and Size |
| 8   | **Pie Chart**    | Showing parts of a whole with a small number of categories (≤5–6).                  | Comparing many categories or small differences.    | Customer Distribution by Day                  |
| 9   | **Count Plot**   | Counting the frequency of categorical values.                                       | Displaying continuous numerical data.              | Number of Customers by Day                    |
| 10  | **Violin Plot**  | Comparing data distributions and density across categories.                         | Simple comparisons where a box plot is sufficient. | Bill Distribution by Time                     |
