backpropagation

```python
 x=torch.tensor(4.,requires_grad=True)
 print(x)
```

```python
y=x**3
print(y)
```

```python
y.backward()
print(x.grad)
```

```python
l1=[[2.,3.,5.],[6.,5.,11.],[13.,17.,18]]
x=torch.tensor(l1,requires_grad=True)
print(x)
```

```python
y=x**3+x**2
print(y)
```

dy/dx=3x^2+2x

```python
z=y.sum()
print(z)
```

```python
z.backward()
```

```python
print(x.grad)
```
