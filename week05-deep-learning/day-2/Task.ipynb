                 Input Layer
             ┌───────────────┐
             │ x1       x2   │
             └──────┬───┬────┘
                    │   │
          W1,b1     │   │
                    ▼   ▼
            Hidden Layer (2 Neurons)
            ┌───────────────┐
            │ h1       h2   │
            └──────┬───┬────┘
                   │   │
              W2,b2│   │
                   ▼   ▼
             Output Layer
             ┌───────────┐
             │     y     │
             └───────────┘

```python
import numpy as np

np.random.seed(42)
```

```python
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y = np.array([
    [0],
    [1],
    [1],
    [0]
])
```

Initialize the parameter

```python
W1 = np.random.randn(2,2)
b1 = np.zeros((1,2))
```

```python
W2 = np.random.randn(2,1)
b2 = np.zeros((1,1))
```

Sigmoidal function 

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
def sigmoid_derivative(x):
    return x*(1-x)
```

Training loop 

```python
learning_rate = 0.5
epochs = 10000

for epoch in range(epochs):

    # Forward Pass

    hidden_input = np.dot(X,W1) + b1
    hidden_output = sigmoid(hidden_input)

    output_input = np.dot(hidden_output,W2) + b2
    predictions = sigmoid(output_input)

    # Loss

    loss = np.mean((y-predictions)**2)

    # Backpropagation

    output_error = y - predictions
    output_delta = output_error * sigmoid_derivative(predictions)

    hidden_error = np.dot(output_delta,W2.T)
    hidden_delta = hidden_error * sigmoid_derivative(hidden_output)

    # Gradient Descent

    W2 += learning_rate * np.dot(hidden_output.T,output_delta)
    b2 += learning_rate * np.sum(output_delta,axis=0,keepdims=True)

    W1 += learning_rate * np.dot(X.T,hidden_delta)
    b1 += learning_rate * np.sum(hidden_delta,axis=0,keepdims=True)

    if epoch % 1000 == 0:
        print(f"Epoch {epoch}, Loss = {loss:.6f}")
```

Final prediction 

```python
print("Predictions")

hidden = sigmoid(np.dot(X,W1)+b1)

output = sigmoid(np.dot(hidden,W2)+b2)

print(output)
```

Probabilty to binary 

```python
binary = (output > 0.5).astype(int)

print(binary)
```
