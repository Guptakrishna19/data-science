Import numpy

```python
import numpy as np
```

Dataset 

```python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([[0], [1], [1], [1]])
```

Weights and bias

```python
np.random.seed(42)

weights = np.random.randn(2, 1)
bias = np.random.randn(1)

print("Weights:\n", weights)
print("Bias:", bias)
```

Implement sigmoid 


```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

y=\frac{1}{1+e^{-x}}

```python
z = np.dot(X, weights) + bias
y_pred = sigmoid(z)

print("Predictions:")
print(y_pred)
```

Mean sq error

```python
loss = np.mean((y - y_pred) ** 2)

print("Loss:", loss)
```

Compute gradient 

```python
sigmoid_derivative = y_pred * (1 - y_pred)
```

```python
dL_dy = -2 * (y - y_pred) / len(y)
```

```python
delta = dL_dy * sigmoid_derivative
```

```python
dw = np.dot(X.T, delta)
```

```python
db = np.sum(delta)
```

```python
print("Gradient Weights:\n", dw)
print("Gradient Bias:", db)
```

Perform gradient 

```python
learning_rate = 0.5
```

```python
weights -= learning_rate * dw
bias -= learning_rate * db
```

```python
print("Updated Weights:\n", weights)
print("Updated Bias:", bias)
```

Forward pass again 

```python
z = np.dot(X, weights) + bias
y_pred_new = sigmoid(z)

new_loss = np.mean((y - y_pred_new) ** 2)

print("New Predictions:")
print(y_pred_new)

print("New Loss:", new_loss)
```

Train for 1000 epochs

```python
epochs = 1000

for epoch in range(epochs):

    z = np.dot(X, weights) + bias
    y_pred = sigmoid(z)

    loss = np.mean((y - y_pred) ** 2)

    dL_dy = -2 * (y - y_pred) / len(y)
    delta = dL_dy * y_pred * (1 - y_pred)

    dw = np.dot(X.T, delta)
    db = np.sum(delta)

    weights -= learning_rate * dw
    bias -= learning_rate * db

    if epoch % 100 == 0:
        print(f"Epoch {epoch}: Loss = {loss:.4f}")
```

Final Prediction 

```python
predictions = sigmoid(np.dot(X, weights) + bias)

print(predictions)
```

```python
binary_predictions = (predictions >= 0.5).astype(int)

print(binary_predictions)
```
