| Scenario                  | Problem Type   | Best Metric   | Justification                                                                                                                                                            |
| ------------------------- | -------------- | ------------- | ------------------------------------------------------------------------------------------------------------------------------------------------------------------------ |
| Fraud Detection           | Classification | **Precision** | Fraud is rare, and false positives can unnecessarily block legitimate customer transactions. High precision ensures that most flagged transactions are truly fraudulent. |
| Customer Churn Prediction | Classification | **F1 Score**  | Businesses want to identify customers likely to leave while avoiding unnecessary retention efforts. F1 Score balances Precision and Recall.                              |
| Medical Diagnosis         | Classification | **Recall**    | Missing a patient with a disease can have severe consequences. High Recall minimizes false negatives by detecting as many actual patients as possible.                   |
| House Price Prediction    | Regression     | **RMSE**      | House prices are continuous values. RMSE penalizes large prediction errors, making it suitable when significant pricing mistakes are costly.                             |
